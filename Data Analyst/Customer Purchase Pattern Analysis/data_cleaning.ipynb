## The dataset used here is UCI ML Repository Online Retail Dataset

##### You can download it from here --> [click here](https://archive.ics.uci.edu/dataset/352/online+retail)

---

```python
import pandas as pd

df = pd.read_excel("Online Retail.xlsx")
print(df.shape)
```

```python
df.head()
```

```python
df.isna().sum()
```

### NOTE: A missing Description almost always corresponds to transactions where the UnitPrice is also 0.

**Checking whether this is true or not**

---

```python
df[df["Description"].isna()].head()
```

```python
if df["Description"].isna().sum() and (df['UnitPrice'] == 0).sum():
    print("Yes")
```

```python
mask = df['Description'].isna() & (df['UnitPrice'] == 0)
new_df = df.drop(index=df[mask].index)
```

```python
new_df.isna().sum()
```

### Checking whether the rows with `UnitPrice==0` has removed or not

---

```python
new_df[new_df['UnitPrice'] == 0].head()
```

```python
new_df[new_df['UnitPrice'] == 0].shape
```

```python
new_df.drop(index=new_df[new_df['UnitPrice'] == 0].index, inplace=True)
```

```python
new_df[new_df['UnitPrice'] == 0].head()
```

### `CustomerID` has `133626` null values
***This is a very high number of missing values. These represent "Guest Checkouts"—people who bought items without logging into an account.***

---

```python
new_df = new_df.fillna("Guest_Customer")

new_df[new_df['CustomerID'] == "Guest_Customer"].head()
```

```python
print(new_df.shape)
new_df.isna().sum()
```
