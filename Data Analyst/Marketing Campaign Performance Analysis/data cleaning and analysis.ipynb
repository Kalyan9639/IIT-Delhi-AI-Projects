```python
import pandas as pd

df = pd.read_csv("marketing_campaign_dataset.csv")
df.head()
```

### Data Cleaning and Analysis

---

```python
print(df.shape)
df.isna().sum()
```

```python
df.info()
```

```python
# 1. Clean Acquisition_Cost: Remove '$' and ',' and convert to float
df['Acquisition_Cost'] = df['Acquisition_Cost'].replace('[\$,]', '', regex=True).astype(float)

# 2. Clean Date: Convert to datetime
df['Date'] = pd.to_datetime(df['Date'])

# 3. Verification: Check the types
print(df.info())
```

### Feature Engineering

---

```python
# Create the calculated metrics
df['Total_Conversions'] = (df['Conversion_Rate'] * df['Clicks']).round(0)
df['Revenue'] = (df['ROI'] * df['Acquisition_Cost']) + df['Acquisition_Cost']
df['Net_Profit'] = df['Revenue'] - df['Acquisition_Cost']

# Check if the numbers look logical
print(df[['Acquisition_Cost', 'Revenue', 'Net_Profit', 'ROI']].head())
```

```python
df.head()
```

### Checking whether the numerical cols have negative values or not

---

```python
numerical_cols = df.select_dtypes(include='number').columns
negative_counts = (df[numerical_cols] < 0).sum()
display(negative_counts)
```
