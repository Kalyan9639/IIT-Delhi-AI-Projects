```python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
from matplotlib.gridspec import GridSpec
import os
import glob

# Set up the figure with a clean style
plt.style.use('seaborn-v0_8-darkgrid')
fig = plt.figure(figsize=(18, 10))
fig.suptitle('E-Commerce Sales Insights Dashboard', fontsize=24, fontweight='bold', y=0.97)

# Create a grid layout: 2 rows x 3 columns for better use of space
gs = GridSpec(2, 3, figure=fig, hspace=0.15, wspace=0.03, 
              left=0.03, right=0.97, top=0.92, bottom=0.08)

# Find all PNG files in the @outputs folder
outputs_path = 'Outputs'
if not os.path.exists(outputs_path):
    # Try alternative path
    outputs_path = '/Workspace/Shared/E-Commerce Sales Insights Dashboard/@Outputs'

image_files = sorted(glob.glob(os.path.join(outputs_path, '*.png')))

print(f"Found {len(image_files)} images in {outputs_path}")
for img_file in image_files:
    print(f"  - {os.path.basename(img_file)}")

# Define positions for each subplot (2 rows x 3 columns)
positions = [
    gs[0, 0],  # Top left
    gs[0, 1],  # Top center
    gs[0, 2],  # Top right
    gs[1, 0],  # Bottom left
    gs[1, 1],  # Bottom center
    gs[1, 2],  # Bottom right
]

# Load and display each image
for idx, img_file in enumerate(image_files[:6]):  # Limit to 6 images
    ax = fig.add_subplot(positions[idx])
    
    # Load and display image
    img = mpimg.imread(img_file)
    ax.imshow(img)
    ax.axis('off')  # Hide axes
    
    # Add a subtle border
    for spine in ax.spines.values():
        spine.set_edgecolor('#cccccc')
        spine.set_linewidth(1)

# Adjust layout
plt.tight_layout()

# Save the dashboard as PNG
dashboard_png_path = os.path.join(outputs_path, 'dashboard.png')
fig.savefig(dashboard_png_path, bbox_inches='tight', dpi=150)

# # Display the dashboard
# display(fig)

print(f"\n✅ Dashboard created successfully and saved as {dashboard_png_path}!")
```
