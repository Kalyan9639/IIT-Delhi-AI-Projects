### Data Cleaning and Feature Engineering:

---

```python
def clean_and_prepare_data(df):
    # 1. REMOVAL
    cols_to_drop = ['Row ID', 'Country/Region']
    df = df.drop(columns=cols_to_drop)
    
    # Remove duplicates
    df = df.drop_duplicates()
    
    # 2. CONVERSION
    df['Order Date'] = pd.to_datetime(df['Order Date'])
    df['Ship Date'] = pd.to_datetime(df['Ship Date'])
    
    # 3. FEATURE ENGINEERING
    # Processing Time
    df['Processing Days'] = (df['Ship Date'] - df['Order Date']).dt.days
    
    # Time-based features
    df['Order Year'] = df['Order Date'].dt.year
    df['Order Month'] = df['Order Date'].dt.month
    df['Order Quarter'] = df['Order Date'].dt.quarter  # Added Quarter
    
    # Profit Margin
    df['Profit Margin %'] = (df['Profit'] / df['Sales']) * 100
    df['Profit Margin %'] = df['Profit Margin %'].fillna(0)
    
    # Order Value Category (Low, Medium, High based on Sales quantiles)
    # Using qcut to bin into 3 equal-sized groups
    df['Order Value Category'] = pd.qcut(df['Sales'], q=3, labels=['Low', 'Medium', 'High'])
    
    print("Preprocessing Complete.")
    print(f"New Columns Created: ['Processing Days', 'Order Year', 'Order Month', 'Order Quarter', 'Profit Margin %', 'Order Value Category']")
    return df
```

```python
df_cleaned = clean_and_prepare_data(df)

print(df_cleaned.shape)
df_cleaned.head()
```
