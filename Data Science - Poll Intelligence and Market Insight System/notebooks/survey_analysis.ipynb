## 3. Brand Preference Analysis

```python
# Initialize analyzer
analyzer = Analyzer(data)

# Brand distribution
brand_dist = analyzer.compute_distribution('brand_label')
print("Brand Distribution (%):")
for brand, pct in brand_dist.items():
    print(f"  {brand}: {pct:.1f}%")
```

```python
# Initialize visualizer
viz = Visualizer(data)

# Plot brand distribution
viz.plot_brand_distribution()
plt.show()
```

## 5. Geographic Analysis

```python
# Brand preference by region
viz.plot_preference_by_region()
plt.show()
```

```python
# Regional segment analysis
region_segments = analyzer.segment_analysis('region_label')

print("Regional Analysis:")
print("="*50)
for seg in region_segments:
    print(f"\n{seg.segment_name}:")
    print(f"  Size: {seg.segment_size:,}")
    print(f"  Dominant Brand: {seg.dominant_brand} ({seg.dominant_share:.1f}%)")
    print(f"  Avg Salary: ${seg.avg_salary:,.0f}")
    print(f"  Avg Age: {seg.avg_age:.1f}")
```

## 8. Insight Generation

```python
# Generate insights
insight_engine = InsightEngine(data, analyzer)
insights = insight_engine.generate_all_insights()

print("KEY BUSINESS INSIGHTS")
print("="*60)

for i, insight in enumerate(insights, 1):
    print(f"\n[{i}] {insight.category}: {insight.title}")
    print(f"    Finding: {insight.finding}")
    print(f"    Implication: {insight.implication}")
    print(f"    Confidence: {insight.confidence}")
```

```python
# Export insights
insights_df = insight_engine.export_insights()
insights_df
```
